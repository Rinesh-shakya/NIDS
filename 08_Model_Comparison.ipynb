```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
comparison = pd.DataFrame({

    "Model":[
        "Naive Bayes",
        "Random Forest",
        "XGBoost",
        "LSTM"
    ],

    "Accuracy":[
        0.714223659,
        0.998552047,
        0.998792050,
        0.986194858774992
    ],

    "Precision":[
        0.976666471,
        0.998527632,
        0.998789382,
        0.9854953311754697
    ],

    "Recall":[
        0.714223659,
        0.998552047,
        0.998792050,
        0.986194858774992
    ],

    "F1 Score":[
        0.815672668,
        0.998521589,
        0.998786523,
        0.9855160643466306
    ],

    "Training Time (s)":[
        5.568719625,
        273.6565347,
        443.7881746,
        435.63
    ],

    "Prediction Time (s)":[
        10.97443891,
        7.888166904,
        4.386807919,
        38.73
    ]

})
```

```python
print(comparison)
```

```python
comparison.to_csv(
    "results/final_model_comparison_withSmote.csv",
    index=False
)
```

```python
plt.figure(figsize=(8,5))   #accuracy graph
plt.bar(
    comparison["Model"],
    comparison["Accuracy"]
)
plt.ylabel("Accuracy")
plt.title("Accuracy Comparison")
plt.xticks(rotation=15)
plt.tight_layout()
plt.savefig("plots/accuracy_comparison.png")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.bar(
    comparison["Model"],
    comparison["Precision"]
)
plt.ylabel("Precision")
plt.title("Precision Comparison")
plt.xticks(rotation=15)
plt.tight_layout()
plt.savefig("plots/precision_comparison.png")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.bar(
    comparison["Model"],
    comparison["Recall"]
)
plt.ylabel("Recall")
plt.title("Recall Comparison")
plt.xticks(rotation=15)
plt.tight_layout()
plt.savefig("plots/recall_comparison.png")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.bar(
    comparison["Model"],
    comparison["F1 Score"]
)
plt.ylabel("F1 Score")
plt.title("F1 Score Comparison")
plt.xticks(rotation=15)
plt.tight_layout()
plt.savefig("plots/f1_comparison.png")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.bar(
    comparison["Model"],
    comparison["Training Time (s)"]
)
plt.ylabel("Seconds")
plt.title("Training Time Comparison")
plt.xticks(rotation=15)
plt.tight_layout()
plt.savefig("plots/training_time.png")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.bar(
    comparison["Model"],
    comparison["Prediction Time (s)"]
)
plt.ylabel("Seconds")
plt.title("Prediction Time Comparison")
plt.xticks(rotation=15)
plt.tight_layout()
plt.savefig("plots/prediction_time.png")
plt.show()
```

```python
ranking = comparison.sort_values(
    by="Accuracy",
    ascending=False
)
print(ranking)
```

```python
ranking.to_csv(
    "results/model_ranking2.csv",
    index=False
)
```
