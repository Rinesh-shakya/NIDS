```python
import os
import pandas as pd
from sklearn.model_selection import train_test_split
from imblearn.combine import SMOTETomek
from imblearn.over_sampling import SMOTE
```

```python
X_train=pd.read_csv('data/processed/X_train.csv') #load
X_test=pd.read_csv('data/processed/X_test.csv')
Y_train=pd.read_csv('data/processed/Y_train.csv')
Y_test=pd.read_csv('data/processed/Y_test.csv')
```

```python
Y_train=Y_train.squeeze() #prepare label
Y_test=Y_test.squeeze()
```

```python
train_df=X_train.copy();train_df['Label']=Y_train  #combine
```

```python
sample_size=1000000  #sample
train_sample, _ =train_test_split(
    train_df,
    train_size=sample_size,
    stratify=train_df['Label'],
    random_state=42
)
```

```python
X_train_sample=train_sample.drop('Label',axis=1)  #split
Y_train_sample=train_sample['Label']
```

```python
sampling_strategy = {
    "Heartbleed": 200,
    "Web Attack � Sql Injection": 200,
    "Infiltration": 300,
    "Web Attack � XSS": 500,
    "Web Attack � Brute Force": 800,
    "Bot": 1000,
    "SSH-Patator": 1500,
    "DoS Slowhttptest": 3000,
    "DoS slowloris": 3000,
    "FTP-Patator": 3000,
    "DoS GoldenEye": 5000
}

smote = SMOTE(
    sampling_strategy=sampling_strategy,
    k_neighbors=1,
    random_state=42
)

smote_tomek = SMOTETomek(
    smote=smote,
    random_state=42
)

X_train_balanced, Y_train_balanced = smote_tomek.fit_resample(
    X_train_sample,
    Y_train_sample
)
```

```python
print("Before balancing:")
print(Y_train_sample.value_counts())

print("\nAfter SMOTE + Tomek:")
print(Y_train_balanced.value_counts())
```

```python
os.makedirs('data/sampled', exist_ok=True)
pd.DataFrame(X_train_balanced).to_csv(
    'data/sampled/X_train_balanced.csv',
    index=False
)
pd.Series(Y_train_balanced).to_csv(
    'data/sampled/Y_train_balanced.csv',
    index=False
)
# Save original test set
X_test.to_csv('data/sampled/X_test.csv', index=False)
Y_test.to_csv('data/sampled/Y_test.csv', index=False)
```
