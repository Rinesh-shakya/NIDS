```python
import pandas as pd
import numpy as np
import joblib
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
df=pd.read_csv('data/raw/merged_dataset.csv',low_memory=False) # load
```

```python
df.columns=df.columns.str.strip() #strip
```

```python
df.replace([np.inf, -np.inf],np.nan,inplace=True)  #clean
df.dropna(inplace=True)
df.drop_duplicates(inplace=True) 
```

```python
X=df.drop('Label', axis=1) #split
Y=df['Label']
```

```python
obj=X.select_dtypes(include='object').columns  #Dummies
if len(obj)>0:
    X=pd.get_dummies(X)
```

```python
# Save processed dataset before scaling

processed_df = X.copy()

processed_df["Label"] = Y.values

processed_df.to_csv(
    "data/processed/processed_dataset.csv",
    index=False
)

print("Processed dataset saved successfully!")
print(processed_df.shape)
```

```python
scaler=StandardScaler()    #scale
X_scaled=scaler.fit_transform(X)
joblib.dump(scaler,'models/scaler.pk1')
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(X_scaled,Y,test_size=0.2,stratify=Y,random_state=42)  #Train/test
```

```python
import os
os.makedirs('data/processed',exist_ok=True)  # Save
```

```python
pd.DataFrame(X_train).to_csv("data/processed/X_train.csv",index=False)
```

```python
pd.DataFrame(X_test).to_csv("data/processed/X_test.csv",index=False)
```

```python
pd.DataFrame(Y_train).to_csv("data/processed/Y_train.csv",index=False)
```
