```python
import pandas as pd
import numpy as np
import glob
import joblib
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split
```

```python
files=glob.glob('MachineLearningCSV/MachineLearningCVE/*.csv')
df_list=[]
for f in files:
    df_list.append(pd.read_csv(f,low_memory=False))
```

```python
df=pd.concat(df_list,ignore_index=True)
print(df.shape)
df.to_csv('data/raw/merged_dataset.csv',index=False)
```

```python
print(df.head())
print(df.info())
print(df.isnull().sum())
```

```python
df.replace([np.inf,-np.inf],np.nan,inplace=True)
df.dropna(inplace=True)
df.drop_duplicates(inplace=True)
```

```python
le=LabelEncoder()
df[' Label']=le.fit_transform(df[' Label'])
joblib.dump(le,'models/label_encoder.pk1')
```
