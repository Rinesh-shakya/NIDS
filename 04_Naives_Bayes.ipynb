```python
import pandas as pd
import joblib
import time
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score,classification_report,confusion_matrix
```

```python
X_train=pd.read_csv('data/sampled/X_train_balanced.csv')
X_test=pd.read_csv('data/sampled/X_test.csv')
Y_train=pd.read_csv('data/sampled/Y_train_balanced.csv').squeeze()
Y_test=pd.read_csv('data/sampled/Y_test.csv').squeeze()
```

```python
nb=GaussianNB()  #train
start=time.time()
nb.fit(X_train,Y_train)
training_time=time.time()-start
print(f"Training Time:{training_time:.2f}seconds")
```

```python
joblib.dump(nb,'models/smote_result/naive_bayes.pk2') #save
```

```python
start_pred =time.time()
Y_pred=nb.predict(X_test)
end_pred=time.time()
prediction_time=end_pred-start_pred
print(f"Prediction Time: {prediction_time:.2f} seconds") #predict
```

```python
accuracy=accuracy_score(Y_test,Y_pred)
precision=precision_score(Y_test,Y_pred,average='weighted')
recall=recall_score(Y_test,Y_pred,average='weighted')
f1=f1_score(Y_test,Y_pred,average='weighted')
print(classification_report(Y_test,Y_pred))
print(confusion_matrix(Y_test,Y_pred))
```

```python
results=pd.DataFrame({'Metric':['Accuracy','Precision','Recall','F1','Training Time(s)', 'Prediction Time(s)'],'Value':[accuracy,precision,recall,f1,training_time,prediction_time]})
results.to_csv('models/smote_result/naive_bayes_results.csv',index=False)
```

```python
report = classification_report(
    Y_test,
    Y_pred,
    output_dict=True,
    zero_division=0
)

report_df = pd.DataFrame(report).transpose()

report_df.to_csv(
    "models/smote_result/naive_bayes_classification_report.csv"
)
```
