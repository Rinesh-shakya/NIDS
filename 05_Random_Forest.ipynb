```python
import os
import pandas as pd
import joblib
import time
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score,classification_report,confusion_matrix
```

```python
X_train = pd.read_csv("data/sampled/X_train_balanced.csv")
X_test = pd.read_csv("data/sampled/X_test.csv")
Y_train = pd.read_csv("data/sampled/Y_train_balanced.csv").squeeze()
Y_test = pd.read_csv("data/sampled/Y_test.csv").squeeze()
```

```python
print("X_train:",X_train.shape)
print("X_test:",X_test.shape)
print("Y_train:",Y_train.shape)
print("Y_train:",Y_test.shape)
```

```python
rf=RandomForestClassifier(                        #create model
    n_estimators=200,
    max_depth=None,
    random_state=42,
    n_jobs=-1
)
```

```python
start_train=time.time()
rf.fit(X_train,Y_train)
end_train=time.time()
training_time=end_train-start_train
print(f"Training Time: {training_time:.2f} seconds")
```

```python
os.makedirs("models",exist_ok=True)
joblib.dump(rf,"models/smote_result/random_forest.pk2")
```

```python
start_pred =time.time()
Y_pred=rf.predict(X_test)
end_pred=time.time()
prediction_time=end_pred-start_pred
print(f"Prediction Time: {prediction_time:.2f} seconds")
```

```python
accuracy = accuracy_score(Y_test, Y_pred)
precision = precision_score(
    Y_test,
    Y_pred,
    average="weighted",
    zero_division=0
)
recall = recall_score(
    Y_test,
    Y_pred,
    average="weighted",
    zero_division=0
)
f1 = f1_score(
    Y_test,
    Y_pred,
    average="weighted",
    zero_division=0
)
print("Accuracy :", accuracy)
print("Precision:", precision)
print("Recall   :", recall)
print("F1 Score :", f1)
```

```python
print(classification_report(Y_test,Y_pred,zero_division=0))
print(confusion_matrix(Y_test,Y_pred))
```

```python
results=pd.DataFrame({'Metric':['Accuracy','Precision','Recall','F1','Training Time(s)', 'Prediction Time(s)'],'Value':[accuracy,precision,recall,f1,training_time,prediction_time]})
results.to_csv('models/smote_result/random_forest_results.csv',index=False)
```
